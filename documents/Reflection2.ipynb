```python
import pandas as pd
from IPython.display import display  # For better table formatting 
# Superheroes DataFrame
superheroes = pd.DataFrame({
    "name": ["Magneto", "Storm", "Mystique", "Batman", "Joker", "Catwoman", "Hellboy"],
    "alignment": ["bad", "good", "bad", "good", "bad", "bad", "good"],
    "gender": ["male", "female", "female", "male", "male", "female", "male"],
    "publisher": ["Marvel", "Marvel", "Marvel", "DC", "DC", "DC", "Dark Horse Comics"]
})

# Publishers DataFrame
publishers = pd.DataFrame({
    "publisher": ["DC", "Marvel", "Image"],
    "year_founded": [1934, 1939, 1992]
})
```

## Different Merge Types
We test different types of merges:

Inner Join: Keeps only matching publishers.
Left Join: Keeps all superheroes, adding publisher info where available.
Right Join: Keeps all publishers, adding superheroes where available.
Outer Join: Keeps all superheroes and publishers, filling missing values with NaN.

```python
# 1 Different Merge Types
inner_merge = pd.merge(superheroes, publishers, on="publisher", how="inner")
left_merge = pd.merge(superheroes, publishers, on="publisher", how="left")
right_merge = pd.merge(superheroes, publishers, on="publisher", how="right")
outer_merge = pd.merge(superheroes, publishers, on="publisher", how="outer")
```

## Using Indicator to Track Merge Source¶
This merge adds a _merge column showing whether a row is from:

"both" → Found in both datasets.
"left_only" → Found only in the superheroes dataset.
"right_only" → Found only in the publishers dataset.

```python
# 2 Using Indicator to Track Merge Source
merge_with_indicator = pd.merge(superheroes, publishers, on="publisher", how="outer", indicator=True)
```

## Using left_on= and right_on= (Merging on Different Column Names)¶
Sometimes, column names don’t match between DataFrames.
Here, we rename "publisher" in the publishers DataFrame to "company" and merge.

```python
# 5 Using Left and Right On (Different Column Names)
publishers_renamed = publishers.rename(columns={"publisher": "company"})
merge_with_diff_names = pd.merge(superheroes, publishers_renamed, left_on="publisher", right_on="company")
```

## Merging Based on Index
Instead of merging on a column, we merge using the index (name in superheroes and publisher in publishers). Using left_on= and right_on= (Merging on Different Column Names)** Sometimes, column names don’t match between DataFrames.
Here, we rename "publisher" in the publishers DataFrame to "company" and merge.

```python
# 6 Merging Based on Index
merge_on_index = pd.merge(superheroes.set_index("name"), publishers.set_index("publisher"), left_index=True, right_index=True, how="inner")
```

# **3. Handling a Compound Key in `pd.merge()`**
A **compound key** is a combination of **two or more columns** used to merge datasets.  
For example, if we have **Iron Man** in both **Comics** and **Movies**, merging only on `"name"` would cause incorrect matches.

---

### Why Use a Compound Key?
 Prevents **wrong merges** when a single column isn't unique.  
 Useful when the same entity appears in **different categories** (e.g., Movies vs. Comics).  
 Helps **differentiate versions of the same entity**.

---

###  Example: Merging Superheroes with Medium (Comics, Movies, TV)
We merge on **both** `"name"` and `"medium"` (Movie/TV/Comic).

```python
import pandas as pd
from IPython.display import display  # For Jupyter formatting

# Superheroes dataset with an added "medium" column
superheroes = pd.DataFrame({
    "name": ["Iron Man", "Iron Man", "Batman", "Superman"],
    "medium": ["Comics", "Movies", "Comics", "TV"],
    "publisher": ["Marvel", "Marvel", "DC", "DC"]
})

# Publishers dataset with additional details
publishers = pd.DataFrame({
    "publisher": ["Marvel", "Marvel", "DC", "DC"],
    "medium": ["Comics", "Movies", "Comics", "TV"],
    "year_founded": [1939, 2008, 1934, 1952]
})

# Merge on both "publisher" and "medium" to differentiate versions
merged_compound_key = pd.merge(superheroes, publishers, on=["publisher", "medium"], how="inner")

# Display merged data
display(merged_compound_key)
```

# **4. Difference Between `join()` and `merge()`**
`pd.merge()` and `df.join()` both combine DataFrames, but they work differently.

- **`pd.merge()`** is more flexible and allows merging on **specific columns**.
- **`df.join()`** is used for merging based on **index**.
*merge() allows more flexibility with join types and keys.
*join() is faster for merging large datasets on an index.

### **When to Use What?**
**Use `pd.merge()`** when merging **two DataFrames on a specific column**.  
**Use `df.join()`** when merging **on index instead of column names**.


---

# **5. What Should Be the Default Merge Type?**
By default, `pd.merge()` uses `"inner"`, meaning:
- Only **matching rows** from both DataFrames are kept.
- Non-matching rows are **dropped**.

### **Best Merge Type?**
 `"inner"` is useful but may lose important data.  
 `"left"` keeps all records from the main dataset.  
 `"outer"` keeps everything, filling missing values.  
 `"right"` is like `"left"`, but prioritizes the second dataset.  
 `"anti"` (not in `pd.merge()`, but available in SQL) removes matching rows.

**For most cases, `"left"` is the best choice to retain all important data.**
