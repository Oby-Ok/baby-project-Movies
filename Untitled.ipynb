## Question 1
### What happens with each pd.merge() parameter?
- `how="inner"`: Keeps only matching records in both datasets.
- `how="left"`: Keeps all records from the left dataset, adding `NaN` where no match is found.
- `how="right"`: Keeps all records from the right dataset, adding `NaN` where no match is found.
- `how="outer"`: Keeps all records from both datasets, adding `NaN` for missing values.
- `on="Name"`: Specifies the column on which to merge.
- `left_on` & `right_on`: Allows merging on different column names in each dataset.
- `suffixes=("_x", "_y")`: Helps distinguish overlapping column names.

# Question 4: Difference Between join and merge

## Explanation:
### The main difference between `merge()` and `join()` in pandas is how they operate:

### - `merge()` is more flexible and works like SQL joins.
### - `join()` is primarily for index-based joining.

### Example 1: Using merge() for joining on a specific column (like SQL JOIN)

````python
import pandas as pd

df1 = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']})
df2 = pd.DataFrame({'ID': [1, 2, 4], 'Age': [25, 30, 40]})

merged_df = pd.merge(df1, df2, on="ID", how="inner")
print("Merge Example:\n", merged_df)

# Example 2: Using join() for index-based joining
df1.set_index('ID', inplace=True)
df2.set_index('ID', inplace=True)

joined_df = df1.join(df2, how="inner")
print("Join Example:\n", joined_df)
````


## Question 5: Default Merge Type**  

### Best Default Merge Type:
- `"inner"` is the safest default since it ensures **only valid matches** are kept.
- `"outer"` is useful when **preserving all data** is necessary.
